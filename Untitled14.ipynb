```python
lt = []
n = int(input("Enter the number of elements: "))
for i in range(n):
    ele = int(input("Enter an element: "))
    lt.append(ele)

# Sort the list in increasing order
lt.sort()

# Split the list into two halves: the first half will remain sorted (ascending),
# the second half will be reversed (descending)
first_half = lt[:(n//2)]  # First half (ascending order)
second_half = lt[(n//2):]  # Second half
second_half.sort(reverse=True)  # Reverse second half (descending order)

# Create the final result by combining the two halves
n_lt = first_half + second_half

print("Rearranged List:", n_lt)
```

```python
#5
lt = []
n = int(input("Enter the number of elements: "))

for i in range(n):
    ele = int(input("Enter: "))
    lt.append(ele)

lt.sort()  # Sort in increasing order

# Splitting into two halves
s_lt = lt[n//2:]  # Second half (larger elements)
s_lt.sort(reverse=True)  # Sort in descending order

n_lt = lt[:n//2]  # First half remains in increasing order
n_lt.extend(s_lt)  # Append second half (already sorted in descending order)

print("Rearranged list:",n_lt)
```

```python
#Kadane's Algorithm 7
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        maxsum=float('-inf')
        for i in range(len(nums)):
            cursum=0
            for j in range(i,(len(nums))):
                cursum+=nums[j]
                if cursum>maxsum:
                    maxsum=cursum
        return maxsum
```

```python
#Stock Buy and Sell 8
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        buy=prices[0]
        pro=0
        for i in range(len(prices)):
            if prices[i]<buy:
                buy=prices[i]
            elif prices[i]-buy>pro:
                pro=prices[i]-buy
        return pro
```

```python
#13
import math

def find_lcm(a, b):
    return (a * b) // math.gcd(a, b)  # Using LCM formula: LCM(a, b) = (a * b) / GCD(a, b)

# Example usage
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("LCM of", num1, "and", num2, "is:", find_lcm(num1, num2))
```

```python
n = int(input("Enter a number: "))
lt = []

# Step 1: Find all prime numbers up to n
for i in range(2, n):  # Start from 2 since 1 is not prime
    c = 0
    for j in range(1, i+1):
        if i % j == 0:
            c += 1
    if c == 2:  # If prime, add to the list
        lt.append(i)

# Step 2: Check if any two prime numbers sum to n
yes = 0
for ele in lt:
    for m in lt:
        if ele + m == n:  # Check sum directly
            yes = 1
            break  # Exit inner loop if found
    if yes == 1:
        break  # Exit outer loop as well

# Step 3: Print result
if yes == 1:
    print("True")
else:
    print("False")
```

```python
text = input("Enter: ")  # Don't use 'str' as a variable name
meow = {}  # Use dictionary to store frequency

# Count occurrences of each letter
for letter in text:
    if letter in meow:
        meow[letter] += 1
    else:
        meow[letter] = 1

print(meow)  # Print the frequency dictionary

# Find the character with the maximum frequency
max_letter = max(meow, key=meow.get)  # Find the key with the highest value
max_count = meow[max_letter]  # Get the maximum count

print(f"'{max_letter}' appears {max_count} times")
```

```python
def remove_duplicates(text):
    result = ""  # String to store the final result
    seen = set()  # Set to keep track of characters that are already in the result

    for letter in text:
        if letter not in seen:
            result += letter  # Add the character to the result
            seen.add(letter)  # Mark it as seen

    return result

# Input from user
text = input("Enter a string: ")

# Call the function and display the result
result = remove_duplicates(text)
print(f"String without duplicates: {result}")

```
